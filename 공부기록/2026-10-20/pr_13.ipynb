# for / while 반복문

- **for** : 묶음을 순서대로 하나씩 (`range`, list, str, dict…)
- **while** : 조건이 참인 동안
- **break** : 즉시 종료 / **continue** : 이번만 건너뛰기

## 1) for + range

- `range(n)` → 0 ~ n-1
- `range(a, b)` → a ~ b-1
- `range(a, b, s)` → a부터 s씩, b 직전까지

```python
for i in range(5):
    print(f"i = {i}")
```

```python
print("1부터 5까지:")
for n in range(1, 6):
    print(n, end=" ")
print()
```

```python
print("짝수만 (2, 4, 6, 8, 10):")
for n in range(2, 11, 2):
    print(n, end=" ")
print()
```

## 2) for + 리스트 / 문자열 / 딕셔너리

```python
fruits = ["사과", "바나나", "포도"]
for f in fruits:
    print(f"과일: {f}")
```

```python
for ch in "hi":
    print(f"글자: {ch}")
```

```python
scores = {"국어": 90, "영어": 85, "수학": 95}
for subject, score in scores.items():
    print(f"{subject} = {score}")
```

```python
# 인덱스도 같이 필요할 때
fruits = ["사과", "바나나", "포도"]
for idx, f in enumerate(fruits):
    print(f"[{idx}] {f}")
```

## 3) while

조건을 바꾸는 줄(`count += 1`)을 빼면 **무한 루프**가 됩니다.

```python
count = 0
while count < 3:
    print(f"count = {count}")
    count += 1
```

```python
total = 0
n = 1
while n <= 5:
    total += n
    n += 1
print(f"1~5 합 = {total}")
```

## 4) break / continue

- `continue` : 이번 회차만 건너뛰고 다음으로
- `break` : 반복 전체 종료

```python
for i in range(10):
    if i == 3:
        continue
    if i == 7:
        break
    print(i, end=" ")
print()
```

```python
nums = [2, 5, 0, 8]  # 0이 나오면 중단
i = 0
while i < len(nums):
    if nums[i] == 0:
        print("0 만남 -> 중단")
        break
    print(f"처리: {nums[i]}")
    i += 1
```

## 5) 중첩 for (구구단 2단)

```python
for dan in range(2, 3):
    for mul in range(1, 10):
        print(f"{dan} x {mul} = {dan * mul}")
```

## 6) for vs while (같은 일)

- **for** : 돌릴 대상이 분명할 때 (`range`, 리스트…)
- **while** : 끝 조건이 중요할 때

```python
for i in range(3):
    print("for:", i)
```

```python
i = 0
while i < 3:
    print("while:", i)
    i += 1
```

## 정리

| 문법 | 의미 |
|------|------|
| `for x in 묶음` | 원소를 순서대로 |
| `for i in range(n)` | 0 .. n-1 |
| `while 조건` | 조건이 True인 동안 |
| `break` / `continue` | 종료 / 이번만 스킵 |
