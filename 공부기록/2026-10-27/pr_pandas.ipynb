```python
from pathlib import Path

downloads = Path.home() / "Downloads"
csv_files = list(downloads.glob("*.csv"))

if not csv_files:
    raise FileNotFoundError(f"CSV 파일을 찾을 수 없습니다: {downloads}")

csv_path = max(csv_files, key=lambda path: path.stat().st_mtime)

try:
    data = pd.read_csv(csv_path, encoding="utf-8-sig")
except UnicodeDecodeError:
    data = pd.read_csv(csv_path, encoding="cp949")

json_path = csv_path.with_suffix(".json")
data.to_json(json_path, orient="records", force_ascii=False, indent=2)

print(f"읽은 파일: {csv_path}")
print(f"저장한 파일: {json_path}")
```
