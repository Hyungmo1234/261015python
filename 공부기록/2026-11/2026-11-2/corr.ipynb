```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Age': [25, 40, 35, 30, 45, 23, 50, 28, 33, 38],
    'Income': [50000, 60000, 55000, 70000, 80000, 45000, 90000, 48000, 52000, 75000]
})

coorrelation = data[['Age','Income']].corr()


plt.figure(figsize=(8,6))
sns.scatterplot(x='Age', y='Income', data=data)
plt.title('Scatter Plot of Age vs Income')
plt.show()
```
