# for — VS Code Python 실습

설명은 Markdown 셀, 예제는 실행 가능한 Python 코드 셀입니다. Python 3.10 이상 권장.

## 01. 리스트 순회

리스트 요소를 순서대로 처리합니다.

```python
for item in ["one","two","three"]:
    print(item, type(item), item+"s")
    print(item)
```

## 02. 튜플 순회

튜플의 각 원소를 차례로 받습니다.

```python
pairs=[(1,2),(3,4),(5,6)]
for a,b in pairs:
    print(pairs)
```

## 03. 문자열 순회

문자열을 글자 단위로 순회합니다.

```python
for ch in "Python":
    print(ch)
```

## 04. range

range는 정수 범위를 생성합니다.

```python
for i in range(1,6):
    print(i)
```

## 05. 합계 구하기

for와 누적 변수를 활용합니다.

```python
total=0
for i in range(1,11):
    total+=i
print(total)
```

## 06. continue

조건에 해당하는 회차를 건너뜁니다.

```python
for i in range(1,11):
    if i%2==0:
        continue
    print(i)
```

## 07. break

원하는 조건을 찾으면 반복을 중단합니다.

```python
for i in range(1,11):
    if i==5:
        break
    print(i)
```

## 08. 중첩 for

구구단 일부를 출력합니다.

```python
for dan in range(2,5):
    for n in range(1,4):
        print(f"{dan} x {n} = {dan*n}")
```

## 09. 리스트 컴프리헨션

반복과 조건으로 새 리스트를 간결하게 생성합니다.

```python
squares=[x*x for x in range(1,6)]
evens=[x for x in range(10) if x%2==0]
print(squares,evens)
```

## 10. enumerate와 실습

인덱스와 값을 함께 출력합니다.

```python
fruits=["사과","배","귤"]
for idx,fruit in enumerate(fruits,start=1):
    print(idx,fruit)
```
