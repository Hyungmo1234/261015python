# if — VS Code Python 실습

설명은 Markdown 셀, 예제는 실행 가능한 Python 코드 셀입니다. Python 3.10 이상 권장.

## 01. if와 else

조건의 참·거짓에 따라 실행 흐름을 나눕니다.

```python
money=True
if money:
    print("택시")
else:
    print("도보")
```

## 02. 들여쓰기와 비교

블록은 공백 4칸으로 들여씁니다.

```python
money=2000
if money>=3000:
    print("택시")
else:
    print("도보")
```

## 03. 논리 연산

and, or, not은 여러 조건을 결합합니다.

```python
money=2000
card=True
if money>=3000 or card:
    print("택시 이용 가능")
```

## 04. in과 not in

자료에 특정 값이 포함되어 있는지 검사합니다.

```python
pocket=["paper","money","phone"]
print("money" in pocket)
print("card" not in pocket)
```

## 05. pass

조건을 만족하더라도 아무 작업도 하지 않을 수 있습니다.

```python
if "money" in ["money","phone"]:
    pass
else:
    print("카드 확인")
print("다음 단계")
```

## 06. elif

여러 조건 중 첫 번째로 참인 분기를 수행합니다.

```python
score=85
if score>=90:
    grade="A"
elif score>=80:
    grade="B"
else:
    grade="C"
print(grade)
```

## 07. match-case

Python 3.10 이상에서 패턴으로 분기합니다.

```python
grade="B"
match grade:
    case "A": print("탁월")
    case "B": print("우수")
    case _: print("기타")
```

## 08. 연쇄 비교

범위를 1 < x < 10처럼 비교할 수 있습니다.

```python
x=5
print(1<x<10)
print(10<=x<=20)
```

## 09. 조건부 표현식

짧은 조건에 따라 값을 선택합니다.

```python
score=85
result="합격" if score>=60 else "불합격"
print(result)
```

## 10. 실습 문제

성적이 90 이상이면 우수, 60 이상이면 통과, 아니면 재학습을 출력합니다.

```python
score=72
if score>=90:
    print("우수")
elif score>=60:
    print("통과")
else:
    print("재학습")
```
