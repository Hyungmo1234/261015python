# while — VS Code Python 실습

설명은 Markdown 셀, 예제는 실행 가능한 Python 코드 셀입니다. Python 3.10 이상 권장.

## 01. 기본 반복

조건이 참인 동안 반복합니다.

```python
count=0
while count<5:
    count+=1
    print(count)
```

## 02. 반복 종료 조건

반복 변수의 갱신을 잊으면 무한 반복될 수 있습니다.

```python
tree_hit=0
while tree_hit<10:
    tree_hit+=1
print("나무가 넘어갑니다",tree_hit)
```

## 03. 메뉴 반복

입력을 대신한 목록으로 메뉴 선택 과정을 재현합니다.

```python
choices=[1,2,4]
i=0
while i<len(choices):
    choice=choices[i]
    i+=1
    if choice==4:
        print("종료")
        break
    print("선택",choice)
```

## 04. break

조건에 맞으면 반복문을 즉시 종료합니다.

```python
coffee=3
while True:
    if coffee==0:
        print("품절")
        break
    coffee-=1
    print("판매 완료, 남은 수량",coffee)
```

## 05. continue

현재 반복의 나머지를 건너뛰고 다음 반복으로 진행합니다.

```python
a=0
while a<10:
    a+=1
    if a%2==0:
        continue
    print(a)
```

## 06. while-else

break 없이 정상 종료했을 때 else 블록이 실행됩니다.

```python
n=0
while n<3:
    n+=1
else:
    print("정상 종료",n)
```

## 07. 조건과 누적

1부터 10까지 합계를 계산합니다.

```python
n=1
total=0
while n<=10:
    total+=n
    n+=1
print(total)
```

## 08. 안전한 무한 반복

while True는 명확한 break 종료 조건과 함께 사용합니다.

```python
n=0
while True:
    n+=1
    if n==5:
        break
print(n)
```

## 09. 반복문 실습

1~20 중 3의 배수만 출력합니다.

```python
n=0
while n<20:
    n+=1
    if n%3==0:
        print(n)
```

## 10. 카운트다운

5부터 1까지 출력하고 종료 메시지를 표시합니다.

```python
n=5
while n>0:
    print(n)
    n-=1
print("완료")
```
